```python
df.drop('track_id',inplace=True,axis=1)
```

```python
df = df.dropna()
```

```python
df = df.drop(columns=['track_name','artist_name','album_name','genre'])
```

```python
df
```

```python
from sklearn.cluster import KMeans
spt = []
for i in range(1,11):
  data = KMeans(n_clusters=i,init='k-means++',random_state=42)
  data.fit(df)
  spt.append(data.inertia_)
  print(data.inertia_)
```

```python
plt.plot(range(1,11),spt)
plt.xticks(range(1,11))
```

```python
model = KMeans(n_clusters=5,init='k-means++',random_state=42)
model.fit(df)
```

```python
y_pred = model.predict(df)
y_pred
```
