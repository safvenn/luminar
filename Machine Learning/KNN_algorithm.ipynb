## Simple Data set

```python
x=[7,7,3,1]
y=[7,4,4,4]
target=['BAD','BAD','GOOD','GOOD']

features=list(zip(x,y))  ## we use zip() for combining the both values like [(7,7),(7,4),(3,4),(1,4)]
features
```

```python
model=KNeighborsClassifier(n_neighbors=3) # create a variable callled model for our ml model usiing KNeighborClassifier object , in KNeighborclassifier we must give N_neighbors , if we dont give it takes 5 as default value
model.fit(features,target) ## then we train our model with input feaatures and out put features by 'model.fit(in,out)' .After all we complrtrly build our model usiing KNN
```

```python
print(model.predict([[3,7]])) # we use 'model.predict([[]])' for predict using our model by giving values
```

#Split features

split coopumns into input and out put features

```python
x=df.iloc[:,:-1].values
```

```python
y=df.iloc[:,-1].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(x)
x=scaler.transform(x)
```

## Train data and test data

train_test_split ===> 4 Variables

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
model=KNeighborsClassifier()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
import joblib
joblib.dump(model, "model.pkl")
joblib.dump(scaler,"scaler.pkl")
```
