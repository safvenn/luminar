```python
b1.drop('ssid',axis=1,inplace=True)
```

```python
b1.dtypes
```

```python



# Create one fingerprint per scan timestamp
wifi_features = b1.pivot_table(
    index="deviceTime",
    columns="bssid",
    values="signalStrength",
    aggfunc="mean"
)

# Access point not detected = -110 dBm
wifi_features = wifi_features.fillna(-110)

wifi_features = wifi_features.reset_index()
wifi_features['floor'] = 'B1'
wifi_features.to_csv(
    "b1 clean.csv",
    index=False
)

print(wifi_features.head())
print(wifi_features.shape)
```
