## Simple Data set

```python
x=[7,7,3,1]
y=[7,4,4,4]
target=['BAD','BAD','GOOD','GOOD']

features=list(zip(x,y))  ## we use zip() for combining the both values like [(7,7),(7,4),(3,4),(1,4)]
features
```

```python
model=KNeighborsClassifier(n_neighbors=3) # create a variable callled model for our ml model usiing KNeighborClassifier object , in KNeighborclassifier we must give N_neighbors , if we dont give it takes 5 as default value
model.fit(features,target) ## then we train our model with input feaatures and out put features by 'model.fit(in,out)' .After all we complrtrly build our model usiing KNN
```

```python
print(model.predict([[3,7]])) # we use 'model.predict([[]])' for predict using our model by giving values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=1) #random_state = 42,1,0 this gives best performance
```

##**Scaling / Normalisation**

make all input features in a same range bcoz range of input features almost same , higher difference in the range may leads to false prediction.\
**2 types of scalling**


1.   Standard Scalar
2.   min max scalar



### **Standard scallar**


---
*equation:-*

*z=(x-u)/s*

*x=input* \
*u=mean* \
*s=standard deviation*

---




```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
print(x_train)
```

```python
model=KNeighborsClassifier(n_neighbors=15)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
y_test
```

## **Confution Matrix**

we evaluate our classification model by confution matrix.
eg:- in this data set output have two class label (output category) so matrix is 2*2 matrix.

technical terms in confution matrix.

1.TP (TRUE POSITIVE)  1 ===> 1

2.TN (TRUE NEGATIVE)  0 ===> 0

3.FP (FALSE POSITIVE) 0 ===> 1

4.FN (FALSE NEGATIVE) 1 ===> 0



---
this evalution methods are works by confution matrix:-

1.accuracy score \
2.recall \
3.precitiion \
4.f1 score

###**1. Accuracy score**

*equation = tp+tn/(tp+tn+fp+fn)*

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_test,y_pred)
score
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in range(1, 21):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(x_train, y_train)

    y_pred = model.predict(x_test)

    print(f"K={k}, Accuracy={accuracy_score(y_test, y_pred):.3f}")
```
