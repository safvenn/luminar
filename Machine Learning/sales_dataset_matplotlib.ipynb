```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('/content/company_sales_data.csv')
```

```python
df
```

```python
#1. Read Total profit of all months and show it using a line plot
```

```python
plt.plot(df['month_number'],df['total_profit'],marker='o')
plt.xlabel('Month',color='r')
plt.ylabel("Total Profit",color='r')
plt.xticks(df['month_number'])
plt.yticks([200000,300000,400000,500000])
```

```python
# 2. Get total profit of all months and show line plot with the following Style properties
#  Generated line plot must include following Style properties:
#  Line Style dotted and Line-color should be red
#  Show legend at the lower right location.
#  X label name = Month Number
#  Y label name = Sold units number
#  Add a circle marker.
#  Line marker color as read
#  Line width should be 3
```

```python
plt.figure(figsize=(8,8))
plt.plot(df['month_number'],df['total_profit'],'o:r',ms=6,mec='r',label='Total Profit',linewidth=3)
plt.xlabel('Month number',color='k')
plt.ylabel("Sold units number",color='k')
plt.legend(loc='lower right')
plt.xticks(df['month_number'])
plt.yticks([200000,300000,400000,500000])
plt.tight_layout()
```

```python
# 3. Read all product sales data and show it  using a multiline plot
#     square marker with color=green
```

```python
df.columns
```

```python
plt.plot(df['month_number'],df['bathingsoap'],marker='s',mfc='g')
plt.plot(df['month_number'],df['facewash'],marker='s',mfc='g')
plt.plot(df['month_number'],df['facecream'],marker='s',mfc='g')
plt.plot(df['month_number'],df['toothpaste'],marker='s',mfc='g')
plt.plot(df['month_number'],df['shampoo'],marker='s',mfc='g')
plt.plot(df['month_number'],df['moisturizer'],marker='s',mfc='g')
plt.xlabel("Month")
plt.ylabel("Sales")

plt.tight_layout()
```

```python
# 4.Read toothpaste sales data of each month and show it using a scatterplot
```

```python
plt.scatter(df['month_number'],df['toothpaste'],marker='o',s=50,edgecolors='r')
plt.xlabel("Month")
plt.ylabel("sales")
```

```python
# 5.Read face cream and facewash product sales data and show it using the bar chart with grid line
```

```python
plt.bar(df['month_number']-0.25,df['facecream'],color='r',label='Face cream',width=0.25,align='edge')
plt.bar(df['month_number']+0.25,df['facewash'],label="Face wash",width=-.25,align='edge')
plt.legend()
plt.grid(alpha=.5)
plt.xlabel("Month")
plt.ylabel("Sales")
```

```python
# 6.Read the total profit of each month and show it using the histogram to see the most common profit ranges
```

```python
plt.hist(df['total_profit'],alpha=0.7,bins=10)
```

```python
# 7.Calculate total sale data for last year for each product and show it using a Pie chart
```

```python
data=df.iloc[:,1:-2].sum()
```

```python
plt.pie(data.values,labels=data.index,autopct="%1.f%%")
plt.legend(loc="lower right")
plt.tight_layout()
```
