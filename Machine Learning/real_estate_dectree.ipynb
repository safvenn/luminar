```python
m=df['X3 distance to the nearest MRT station'].mean()

for i in df.index:
  if df.loc[i,'X3 distance to the nearest MRT station']>3000:
    df.loc[i,'X3 distance to the nearest MRT station']=m

```

```python
w=df['X6 longitude'].mean()

for i in df.index:
  if df.loc[i,'X6 longitude']>121.56 or df.loc[i,'X6 longitude']<=121.50:
    df.loc[i,'X6 longitude']=w
```

```python
sns.boxplot(y=df['X5 latitude'],data=df)
```

```python
me=df['X5 latitude'].mean()

for i in df.index:
  if df.loc[i,'X5 latitude']>25.00 or df.loc[i,'X5 latitude']<24:
    df.loc[i,'X5 latitude']=me
```

```python
x=df.iloc[:,:-1].values
```

```python
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)

```

```python
from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler()

x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeRegressor

model=DecisionTreeRegressor()

model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
df1=pd.DataFrame({'actual value':y_test,'Predicted_value':y_pred,'Difference':y_test-y_pred})
df1
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score

MAPE=mean_absolute_percentage_error(y_test,y_pred)
R2_score=r2_score(y_test,y_pred)
```

Tuning

```python
from sklearn.model_selection import GridSearchCV
```

```python
model1 = DecisionTreeRegressor()
```

```python
# Fix the parameter grid to use regression criteria instead of classification criteria
param_grid = {
    "criterion": ["squared_error", "friedman_mse", "absolute_error", "poisson"],
    "max_depth": [3, 5, 7, 10, 15, None],
    "min_samples_split": [2, 5, 10, 20],
    "min_samples_leaf": [1, 2, 4, 8],
    "max_features": [None, "sqrt", "log2"],
    "splitter": ["best", "random"]
}

# Re-initialize the grid search with the correct parameter grid
grd = GridSearchCV(model1, param_grid, cv=10, scoring='r2')
grd.fit(x_train, y_train)
```

```python
print(grd.best_params_)
```

```python
tuned = DecisionTreeRegressor(criterion='absolute_error',max_depth=3,min_samples_split=8,splitter='best')
```

```python
tuned.fit(x_train,y_train)
```

```python
y_pred2 = tuned.predict(x_test)
```

```python
MAPE=mean_absolute_percentage_error(y_test,y_pred2)
R2_score=r2_score(y_test,y_pred2)
```
