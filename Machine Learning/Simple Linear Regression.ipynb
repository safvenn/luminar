# Using Datasets

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
df = pd.read_csv('/content/Salary_Data.csv')
```

```python
df
```

```python
df.isna().sum()
```

```python
x =  df.iloc[:,:-1]

y = df.iloc[:,-1]
# donot convert into array
```

```python
x
```

```python
y
```

```python
sns.scatterplot(data=df,x=df['YearsExperience'],y=df['Salary'],color='r')
plt.title('Employee Salary',color='r')
```

#train / test split

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression # Import linear Regression
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
#Line Plot

plt.scatter(x=x_train,y=y_train)
plt.plot(x_test,y_pred,color='r')
plt.ylabel("Salary")
plt.xlabel("Year of Experience")
```

```python
# To get slope (m)
print("slope is ",model.coef_)
```

#performance Evaluation Methods
calculate performance by errors

**METHODS**

1.  **MAE** - *MEAN ABSOLUTE ERORR*
2.   **MAPE** -*MEAN ABSOLUTE PERCENTAGE ERORR*
3.  **MSE** -*MEAN SQUARED ERORR*
4.   **RMSE** -*ROOT MEAN SQAURED ERORR*
5.   **R2_SCORE** -*COEFFIENCE OF DETERMINATION*








```python
from sklearn.metrics import mean_absolute_error,mean_absolute_percentage_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
mae = mean_absolute_error(y_test,y_pred)
mape = mean_absolute_percentage_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)



```

```python
print('mae :',mae)
print('mape :',mape)
print('mse :',mse)
print('rmse:',rmse)
print('r2 score :',score)
```
