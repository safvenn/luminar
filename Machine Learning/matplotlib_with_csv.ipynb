```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/content/customer1.csv")
```

```python
df
```

```python
# no of customers in each location
```

```python
loc_count=df.groupby('loc')['loc'].count()
loc_count
```

```python
#create bar chart
```

```python
plt.bar(loc_count.index,loc_count.values,color='g') #.index give us index and .values gives values
plt.xlabel("County")
plt.ylabel("No of Customers")
plt.suptitle("Customers from each locations",size=13)
```

```python
plt.pie(loc_count.values,labels=loc_count.index)
plt.legend()
```

```python
#ins=dia work , each profession count[bar,pie]
```

```python
india = df[df['loc']=='india'].groupby('prof')['prof'].count()
plt.figure(figsize=(10,10))
plt.barh(india.index,india.values)
plt.xlabel("no of cust")
plt.ylabel('profession')
```

```python
plt.figure(figsize=(10,10))
plt.pie(india.values,labels=india.index)
plt.legend()
plt.tight_layout()
```

```python
#uk work eavh professiom count [pie]
```

```python
uk=df[df['loc']=='uk'].groupby('prof')['prof'].count()
plt.figure(figsize=(10,10))
plt.pie(uk.values,labels=uk.index,autopct='%1.f%%')
plt.legend()
```
