````python
dfe = pd.get_dummies(df[['Location', 'Fuel_Type', 'Transmission', 'Owner_Type','Brand']],drop_first=True,dtype='int')
dfe
````

merge both data frames

````python
df1 = pd.concat([df,dfe],axis=1)
````

````python
df1
````

````python
df1.select_dtypes('object').columns
````

````python
df1.drop(['Unnamed: 0','Name', 'Location', 'Fuel_Type', 'Transmission', 'Owner_Type',
  'New_Price', 'Brand'],axis=1,inplace=True)
````

````python
df1.dtypes
````

````python
df1['Power'].unique()
````

````python
# 1. Clean Mileage: remove ' km/kg' and ' kmpl'
df1['Mileage'] = df1['Mileage'].str.replace(' km/kg', '')
df1['Mileage'] = df1['Mileage'].str.replace(' kmpl', '')


# 2. Clean Engine: remove ' CC'
df1['Engine'] = df1['Engine'].str.replace(' CC', '')


# 3. Clean Power: remove ' bhp' and handle 'null' values
df1['Power'] = df1['Power'].str.replace(' bhp', '')


# Display the cleaned columns
display(df1[['Mileage', 'Engine', 'Power']].head())

df1['Mileage'] = df1['Mileage'].str.replace('null', '0')
df1['Power'] = df1['Power'].str.replace('null', '0')
df1['Engine'] = df1['Engine'].str.replace('null', '0')
````

Change datatypes

````python
df1['Mileage'] = df1['Mileage'].astype(float)
df1['Engine'] = df1['Engine'].astype(float)
df1['Power'] = df1['Power'].astype(float)
````

````python
df1.dtypes
````

Now we change the engine == 0 to nan\
now we can handle missing

````python
df1.loc[df1.Mileage == 0,'Mileage'] = np.nan
df1.loc[df1.Engine == 0,'Engine'] = np.nan
df1.loc[df1.Power == 0,'Power'] = np.nan
df1.loc[df1.Seats == 0,'Seats'] = np.nan
````

# Handle outlierss

````python
df1.columns
````

````python
col = ['Year', 'Kilometers_Driven', 'Mileage', 'Engine', 'Power', 'Seats']
for i in col:
  if i not in df1.columns:
    continue
  q3 = df1[i].quantile(.75)
  q1 = df1[i].quantile(.25)
  iqr = q3-q1
  upper = q3+1.5*iqr
  lower = q1-1.5*iqr

  df1.loc[
    (df1[i] > upper) | (df1[i] < lower),
    i
  ] = df1[i].median()
````
