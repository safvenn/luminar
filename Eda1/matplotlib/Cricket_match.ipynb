```python
over=df['over']
runs=df['runs']
wicket=df['wicket']

plt.figure(figsize=(8,6))


plt.bar(over,runs,label='RUNS')



for o,w,r in zip(over,wicket,runs):
    for i in range(w):
      plt.scatter(o , r + 0.5 +(i*1),color='red',s=100)
plt.scatter([],[],color='red',label='wickets')
plt.xticks(over,color='purple')
plt.yticks(range(2,21,2),color='brown')
plt.xlabel("OVERS")
plt.ylabel("RUNS")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
```
