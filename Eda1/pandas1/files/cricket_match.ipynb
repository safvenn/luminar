```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('/content/cricket_team_dataset.csv')
df
```

```python
over=df['over']
runs=df['runs']
wicket=df['wicket']

plt.figure(figsize=(8,6))


plt.bar(over,runs,label='RUNS')
plt.xticks(over,color='purple')
plt.yticks(range(2,21,2),color='brown')



for o, r, w in zip(over, runs, wicket):
    for i in range(w):
        plt.scatter(o, r + 0.5 + (i * 1), color='red', s=100)
plt.scatter([],[],color='r',label='WICKETS')

plt.xlabel("OVERS",color='g')
plt.ylabel("RUNS",color='g')
plt.title("RUNS AND WICKETS PER OVER",color='r')
plt.legend()
plt.grid()
```
