1.perform inner join


```python
import pandas as pd 

df=pd.read_csv(r'C:\Desktop\luminar\Eda1\pandas1\files\custom_windows.csv')
df1=pd.read_csv(r'C:\Desktop\luminar\Eda1\pandas1\files\order_windows.csv')
```

```python
df2=pd.merge(df,df1,on='id',how='inner')
```

2.salary above 2k


```python
salry_abve2=pd.merge(df,df1,on='id',how='inner')[df2['salary']>2000]
print(salry_abve2)
```

3.latst date purchase date , name ,age, loc,amount,date


```python
pd.merge(df,df1,on='id',how='inner').sort_values('dat').head(1)[['name','age','location','dat']]
```

4.age  minimum 1 emp name , age , dat,amount


```python
pd.merge(df,df1,on='id',how='inner').sort_values('age').head(1)[['name','age','dat','amount']]
```

5.each location count

```python
pd.merge(df,df1,on='id',how='inner').groupby('location')['location'].count()
```
